```python
corr_matrix=df.corr(numeric_only=True)
plt.figure(figsize=(12,5))
sns.heatmap(corr_matrix,annot=True , cmap="coolwarm")
plt.show()
```

```python
plt.figure(figsize=(9, 6))
plt.hexbin(df["torque"], df['stator_winding'], gridsize=40, cmap='Reds', mincnt=1)
plt.colorbar(label='Data Point Count')
plt.xlabel('Torque')
plt.ylabel('Stator Winding Temp')
plt.title('Hexbin Density Plot of Torque vs Stator Winding')
plt.show()
```

```python
result= df.groupby("profile_id")["stator_winding"].agg(["min","max","mean","std","count"])
print(result.to_string())
```

```python
l = ["stator_winding","i_d","coolant","motor_speed","ambient","u_d","torque"]
```

```python

for i in l:
    print(i)
    result = df.groupby("profile_id")[i].agg(["min","max","mean","count","std"])
    print(result.to_string())
```

```python
profile_summary=df.groupby("profile_id")[l].mean()
print(profile_summary)
profile_summary.describe()
```

```python
result = profile_summary.sort_values("stator_winding",ascending=False)[l]
print(result.to_string())
```
